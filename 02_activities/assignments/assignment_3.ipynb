```python
#loading libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.express as px
```

```python
# Load dataset and see column headers
df = pd.read_csv(r"C:\Users\laure\visualization\02_activities\energydata.csv")
df.info()
```

```python
df_2 = df[(df['Class of electricity producer'] == 'Total all classes of electricity producer')
][['REF_DATE', 'GEO', 'Class of electricity producer', 'Type of electricity generation', 'VALUE']].copy()

# # Remove unwanted power types ('Total installed capacity' and 'Total thermal').  We can sum the remaining to calculate these.
df_2 = df_2[~df_2['Type of electricity generation'].isin(['Total installed capacity', 'Total thermal'])].reset_index(drop=True)

# #select the ref date as 2023 to hone in the data set
df_2 = df_2[df_2['REF_DATE'] == 2023].reset_index(drop=True)

# Ensure no zero or NaN values are included
df_2 = df_2[df_2['VALUE'] > 0].reset_index(drop=True)

# Convert the 'VALUE' column to MW (divide by 1000)
df_2['VALUE_MW'] = df_2['VALUE'] / 1000
```

```python
# confirm the updates made above work
df_2.info()
print(df_2["GEO"].unique(), df_2["Class of electricity producer"].unique(), df_2["Type of electricity generation"].unique(),df_2["VALUE"].unique())

```

```python
# Create the treemap with a color-blind friendly discrete color palette
fig = px.treemap(df_2,
                 path=['GEO', 'Type of electricity generation',],  # Hierarchy
                 values='VALUE_MW',  # Size of each block
                 color='Type of electricity generation',  # Color by power type (categorical)
                 hover_data={'VALUE_MW': ':.2f'},  # Show MW value on hover (formatted)
                 color_discrete_sequence=px.colors.qualitative.Set3,  # Discrete color palette
                 title="Electricity Generation (MW) by Province and Power Type in 2023")

# Update layout to set the background color to black and customize legend
fig.update_layout(
    width=1400,  # Increased width
    height=900,  # Increased height
    plot_bgcolor='black',  # Background color of the plot area
    paper_bgcolor='black',  # Background color of the whole figure
    font_color='white',  # Font color for text elements
    title_font=dict(size=22, color='white', family='Arial'),  # Bigger, clearer title
)

# Show the treemap
fig.show()
fig.write_html("electricity_generation_treemap.html")
```
