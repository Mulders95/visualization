```python
#import library
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import requests
import scipy
from PIL import Image
from io import BytesIO

# setting random seed and creating x & y variables
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100, 50)
y2 = np.random.randint(0, 100, 50)

```

```python
#creating fig with 2 subplots
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x, y1, label = "Person 1")
ax.plot(x ,y2, label = "Person 2")
ax.legend(loc = 'lower right',
    frameon = True,
    fontsize = 12,
    ncol = 2,
    shadow = True)
```

```python
#anchoring the legend outside plot
fig, ax = plt.subplots(figsize = (5,3))
ax.plot(x,y1, label = 'person1')
ax.plot(x, y2, label = 'person2')
ax.legend(loc='lower left', 
          bbox_to_anchor=(1, 1))
```

```python
#using scatter and adding in text with customization
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1, label = 'person1')
ax.scatter(x, y2, label = 'person2')
ax.legend(loc='lower left')
ax.text(10,95, "this value is important", 
        ha = 'center',
        color = 'red',
        size =20)
```

```python
#plotting data points using reference placements  
fig, ax = plt.subplots(figsize = (5,3))
ax.axis([0,10,0,10])

ax.text(1,5, "data (1,5)", transform = ax.transData)
ax.text(0.5,.1, "axes: (0.5,.1)", transform = ax.transAxes)
ax.text(0.2,.2, "axes: (0.2,.2)", transform = fig.transFigure)
```

```python
#plotting data and anotating using arrow!
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label = "person 1")
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = 'lower right')

ax.annotate("this is important!", xy = (10,95), xytext = (20,94),
arrowprops = dict(arrowstyle = "wedge", color = "hotpink"))
```

```python
#adding in the x & y axis
fig, ax = plt.subplots(figsize = (5,3))
ax.scatter(x,y1,label = "person 1")
ax.scatter(x,y2, label = "person 2")
ax.legend(loc = 'lower right')

#ax.yaxis.set_major_locator(plt.NullLocator())
#ax.xaxis.set_major_locator(plt.NullFormatter())
#ax.xaxis.set_major_locator(plt.MaxNLocator(3))
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
plt.xticks(rotation = 45, ha = "right", fontsize = 12, font = 'serif')
plt.xlabel("shiny new label", loc = 'center', fontsize= 20, font = 'serif', color = 'hotpink')
```

```python
x1 = np.arange(50)
y1 = np.random.randint(0,75,50)
x2 = np.array(["liffy", "luffy", "lauren"])
y2 = np.array([100, 180, 200])
```

```python
fig, (ax1, ax2) = plt.subplots(ncols =2,
                            nrows =1,
                            figsize =(7,3))
ax1.scatter(x1,y1)
plt.title("plot one")
ax2.bar(x2,y2)
plt.title("plot two")
```

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
['ax2','ax3']],
figsize = (7,4),
layout = "constrained")

someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)

someaxes["ax1"].set_xlabel('a big label', fontsize = 20)
someaxes["ax2"].set_xlabel('a bigger label', fontsize = 20)
someaxes["ax3"].set_xlabel('a big bigger massive label', fontsize = 20)
```

```python
# first make our sample data
x = np.array(["Luffy","Zoro", "Nami", "Usopp","Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "indigo")
ax.plot(x, y2,
color = "red")
```

```python
y2_sd = np.std(y2)
fig, ax = plt.subplots(figsize=(7, 3))
ax.plot(x,y2,color ='red')

ax.errorbar(x, y2, yerr=y2_sd,
    fmt="none",
    ecolor="indigo",
    elinewidth=4,
    capsize=6,
    capthick=4
)
```

```python
x= np.array(["luffy", "zorro", "Nami", "Usopp", "sanji"])
y2 = np.array([170,100,90,120,50])

response = requests.get("https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png")
img_file = BytesIO(response.content)
image = Image.open(img_file)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color= "red")

ax_image = fig.add_axes([.1,.11,.15,.35])

ax_image.imshow(image)
ax_image.axis('off')
```
